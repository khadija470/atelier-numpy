## Partie 1 — Listes Python vs tableaux NumPy

### Concept : créer un tableau NumPy

**Objectif :** transformer une liste Python en tableau NumPy pour pouvoir faire des calculs mathématiques directement sur toutes les valeurs.

**Syntaxe :** `np.array(une_liste)`

**Exemple :** voir la case de code ci-dessous.

```python
temperatures_liste = [24.5,25.1,26.3,27.0,25.8]
temperatures_array = np.array(temperatures_liste)
print(temperatures_liste)
print(temperatures_array)
```

```python
print(temperatures_liste * 2)
```

```python
print(temperatures_array * 2)
```

### Observation (question 4)

- Sur une **liste Python**, `* 2` **répète** les éléments → la liste passe de 5 à 10 valeurs.
- Sur un **tableau NumPy**, `* 2` **multiplie chaque valeur par 2** → on garde 5 valeurs, chacune doublée.

Conclusion : NumPy applique les opérations mathématiques à toutes les valeurs d'un coup. C'est la **vectorisation**, l'atout majeur de NumPy.

## Partie 2 — Création des données

Cette partie présente les fonctions NumPy qui créent automatiquement un tableau.

### `np.zeros` — tableau de zéros
- **Objectif :** créer un tableau rempli de zéros.
- **Syntaxe :** `np.zeros(nombre_de_valeurs)`

### `np.ones` — tableau de uns
- **Objectif :** créer un tableau rempli de 1.
- **Syntaxe :** `np.ones(nombre_de_valeurs)`

### `np.full` — tableau rempli d'une valeur choisie
- **Objectif :** créer un tableau rempli d'une même valeur.
- **Syntaxe :** `np.full(nombre_de_valeurs, valeur)`

### `np.eye` — matrice identité
- **Objectif :** créer une matrice identité (matrice carrée avec des 1 sur la diagonale, des 0 ailleurs).
