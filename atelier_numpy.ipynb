## Partie 1 — Listes Python vs tableaux NumPy

**Énoncé** — on dispose des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0, 25.8.
1. Créer une liste Python stockant ces températures.
2. Créer un tableau NumPy équivalent.
3. Effectuer une multiplication par 2 sur ces deux collections.
4. Quelle différence observe-t-on ?

### Concept : créer un tableau NumPy à partir d'une liste
- **Objectif :** transformer une liste Python en tableau NumPy pour faire des calculs directement sur toutes les valeurs.
- **Syntaxe :** `np.array(une_liste)`
- **Exemple :** voir les cellules ci-dessous.

```python
# Q1 : la liste Python
temperatures_liste = [24.5, 25.1, 26.3, 27.0, 25.8]

# Q2 : le tableau NumPy equivalent
temperatures_array = np.array(temperatures_liste)

print(temperatures_liste)
print(temperatures_array)
```

```python
# Q3 : multiplication par 2 des deux collections
print(temperatures_liste * 2)
print(temperatures_array * 2)
```

**Q4 — différence observée :**
- Sur une liste Python, `* 2` répète les éléments : la liste passe de 5 à 10 valeurs.
- Sur un tableau NumPy, `* 2` multiplie chaque valeur par 2 : on garde 5 valeurs, chacune doublée.

NumPy applique donc l'opération à toutes les valeurs d'un coup : c'est la **vectorisation**, son principal atout.

## Partie 2 — Création des données

**Énoncé**
1. Créer un tableau de température de 10 capteurs, initialement à zéro.
2. Créer un tableau de 10 valeurs égales à 1.
3. Créer un tableau de 10 valeurs égales à 25.
4. Créer une matrice identité de taille 5 × 5.
5. Créer les 24 heures (de 0 à 23) d'une journée.
6. Créer 10 valeurs régulièrement réparties entre 20 °C et 30 °C.
7. Créer aléatoirement les températures de 100 mesures, de façon reproductible.

### Concepts (questions 1 à 4)
- **`np.zeros(n)`** — tableau de `n` zéros.
- **`np.ones(n)`** — tableau de `n` uns.
- **`np.full(n, valeur)`** — tableau de `n` fois la même valeur.
- **`np.eye(taille)`** — matrice identité (carrée, des 1 sur la diagonale, des 0 ailleurs).

### Concepts (questions 5 à 7)
- **`np.arange(fin)`** — suite d'entiers de 0 à `fin` exclu.
- **`np.linspace(debut, fin, n)`** — `n` valeurs régulièrement espacées entre `debut` et `fin` (bornes comprises).
- **`np.random.seed(42)` puis `np.random.uniform(min, max, n)`** — `n` valeurs aléatoires reproductibles (la graine fixe le hasard).

```python
# Q1 : 10 capteurs a zero
capteurs_zero = np.zeros(10)
print(capteurs_zero)

# Q2 : 10 valeurs egales a 1
capteurs_un = np.ones(10)
print(capteurs_un)

# Q3 : 10 valeurs egales a 25
capteurs_25 = np.full(10, 25)
print(capteurs_25)

# Q4 : matrice identite 5x5
identite = np.eye(5)
print(identite)
```

```python
# Q5 : les 24 heures (0 a 23)
heures = np.arange(24)
print(heures)

# Q6 : 10 valeurs entre 20 et 30
temperatures_lin = np.linspace(20, 30, 10)
print(temperatures_lin)

# Q7 : 100 mesures aleatoires reproductibles
np.random.seed(42)
mesures_alea = np.random.uniform(20, 30, 100)
print(mesures_alea)
```

## Partie 3 — Exploration des tableaux

**Énoncé**
1. Créer aléatoirement un tableau des 24 températures d'une journée.
2. Déterminer : a) le nombre de dimensions, b) le nombre de valeurs, c) le type des données, d) la forme du tableau.

### Concepts
- **`.ndim`** — nombre de dimensions du tableau.
- **`.size`** — nombre total de valeurs.
- **`.dtype`** — type des données.
- **`.shape`** — forme du tableau (dimensions détaillées).

```python
# Q1 : 24 températures aléatoires (reproductibles) entre 20 et 30
np.random.seed(0)
temperatures_jour = np.random.uniform(20, 30, 24)
print(temperatures_jour)

# Q2a : nombre de dimensions
print(temperatures_jour.ndim)

# Q2b : nombre de valeurs
print(temperatures_jour.size)

# Q2c : type des données
print(temperatures_jour.dtype)

# Q2d : forme du tableau
print(temperatures_jour.shape)
```

## Partie 4 — Indexation et slicing

**Énoncé** — à partir du tableau des 24 températures, afficher :
1. la première mesure
2. la dernière mesure
3. la température de 8h
4. les températures de 8h à 12h
5. les températures des 6 premières heures
6. les températures des heures paires

### Concepts
- **Indexation** `tableau[i]` — une valeur à la position `i` (on compte à partir de 0).
- **Indice négatif** `tableau[-1]` — la dernière valeur.
- **Slicing** `tableau[début:fin]` — une tranche (fin exclue).
- **Pas** `tableau[début:fin:pas]` — un élément tous les `pas`.

```python
# Q1 : première mesure
print(temperatures_jour[0])

# Q2 : dernière mesure
print(temperatures_jour[-1])

# Q3 : température de 8h
print(temperatures_jour[8])

# Q4 : températures de 8h à 12h
print(temperatures_jour[8:12])

# Q5 : les 6 premières heures
print(temperatures_jour[:6])

# Q6 : les heures paires
print(temperatures_jour[::2])
```

## Partie 5 — Filtrage booléen

**Énoncé** — à partir du tableau des 24 températures :
1. Afficher les températures supérieures à 30 °C.
2. Afficher les températures « normales » (entre 20 et 30 °C).
3. Créer `anomalies`, tableau booléen des températures > 35 °C.
4. Déterminer le nombre de températures anormales à l'aide de `anomalies`.

### Concepts
- **Masque booléen** `tableau > valeur` — renvoie un tableau de True/False.
- **Filtrage** `tableau[condition]` — garde les valeurs où la condition est vraie.
- **Combiner** `(cond1) & (cond2)` — ET ; `(cond1) | (cond2)` — OU (parenthèses obligatoires).
- **Compter les True** `np.sum(masque)` — additionne les True (True=1, False=0).

```python
# Q1 : températures supérieures à 30°C
print(temperatures_jour[temperatures_jour > 30])
# Q2 : températures normales (entre 20 et 30°C)
print(temperatures_jour[(temperatures_jour >= 20) & (temperatures_jour <= 30)])
# Q3 : tableau booléen des anomalies (> 35°C)
anomalies = temperatures_jour > 35
print(anomalies)
# Q4 : nombre de températures anormales
print(np.sum(anomalies))

# Test : on force une anomalie pour vérifier comment le code reagit
temperatures_test = temperatures_jour.copy()   # une copie pour ne pas toucher l'original
temperatures_test[5] = 38.0                     # on met 38°C à l'index 5

anomalies_test = temperatures_test > 35
print(anomalies_test)              # un seul True
print(np.sum(anomalies_test))      # affiche 1
```

## Partie 6 — Manipulation des dimensions

**Énoncé** — on suppose 5 capteurs mesurant sur 24 heures.
1. Créer aléatoirement la matrice des températures (5 capteurs × 24 heures).
2. Transformer la matrice en tableau unidimensionnel.
3. Déterminer la transposée de la matrice.
4. Extraire toutes les mesures du capteur 1.
5. Extraire toutes les mesures du capteur 5.
6. Extraire les mesures de tous les capteurs à 10h.
7. Extraire les mesures des capteurs 2 à 4 entre 8h et 12h.

### Concepts
- **Matrice aléatoire** `np.random.uniform(min, max, (lignes, colonnes))`.
- **Aplatir** `matrice.flatten()` — repasse en une seule dimension.
- **Transposer** `matrice.T` — échange lignes et colonnes.
- **Indexation 2D** `matrice[ligne, colonne]`, avec `:` pour « tout ».

```python
# Q1 : matrice 5 capteurs x 24 heures
np.random.seed(1)
matrice = np.random.uniform(20, 30, (5, 24))
print("Forme :", matrice.shape)
print(matrice)
```

```python
# Q2 : aplatir en une seule dimension
print(matrice.flatten())
```

```python
# Q3 : transposee (devient 24 x 5)
print("Forme transposee :", matrice.T.shape)
print(matrice.T)
```

```python
# Q4 : capteur 1 -> ligne d'index 0
print(matrice[0, :])
# Q5 : capteur 5 -> ligne d'index 4
print(matrice[4, :])
# Q6 : tous les capteurs a 10h -> colonne d'index 10
print(matrice[:, 10])
# Q7 : capteurs 2 a 4 (index 1 a 3) entre 8h et 12h (colonnes 8 a 11)
print(matrice[1:4, 8:12])
```
